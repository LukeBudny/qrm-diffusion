```python
from safetensors import safe_open
```

```python
def get_depth_from_ckpt(path):
    with safe_open(path, framework="pt", device="cpu") as f:
        shape = f.get_tensor("model.diffusion_model.x_embedder.proj.weight").shape
        depth = shape[0] // 64
        print(f"{path}: hidden_size = {shape[0]}, depth = {depth}")
```

```python

# Example usage
get_depth_from_ckpt("models/sd3.5_medium.safetensors")
get_depth_from_ckpt("models/sd3.5_large.safetensors")
get_depth_from_ckpt("models/sd3_medium.safetensors")
```
